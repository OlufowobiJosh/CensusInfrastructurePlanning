# Census Population Analysis & Illustrative Community Planning Scenarios

**Public, privacy-preserving recreation** of an updated university census analysis. Uses aggregate summaries only. The original notebook and source CSV are not included because they contain names and addresses. Scenario projections are illustrative, **not validated demographic forecasts**.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
root = Path(".")
age = pd.read_csv(root / "population_by_age_gender.csv")
scenarios = pd.read_csv(root / "illustrative_population_scenarios.csv")
print("Population:", age.population.sum())
```

## Population structure
The original notebook explored data quality, age, gender, household composition, employment, marital status, religion and health-related variables. This public version intentionally limits granularity.

```python
age_total = age.groupby("age_group", sort=False).population.sum()
ax = age_total.plot.bar(figsize=(12,4), title="Population by age group")
ax.set_ylabel("Residents")
plt.tight_layout()
plt.show()
```

## Illustrative future population scenarios
The updated coursework estimates birth and death proxies from a single census snapshot, but **does not supply a longitudinal series, verified vital statistics or calibrated predictive model**. Therefore, the following 5-, 10-, 15- and 20-year numbers are transparent hypothetical growth-rate scenarios, not evidence-based forecasts.

```python
wide = scenarios.pivot(index="years_ahead", columns="scenario", values="projected_population")
ax = wide.plot(marker="o", figsize=(10,5), title="Illustrative population scenarios — not forecasts")
ax.set(xlabel="Years ahead",ylabel="Projected residents")
plt.tight_layout()
plt.show()
wide
```

## Planning interpretation
Local authorities can use scenario ranges to discuss potential future demand for schools, healthcare and housing, but **facility counts cannot be estimated reliably without service utilisation rates, catchment areas, capacity, migration, birth and mortality data**. Do not use these scenarios for operational decisions.
